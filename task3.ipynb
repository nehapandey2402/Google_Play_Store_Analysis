```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Create a Sample Google Play Store Dataset
np.random.seed(42)
n_apps = 100

categories = ['TOOLS', 'ENTERTAINMENT', 'EDUCATION', 'BUSINESS', 'LIFESTYLE', 'FINANCE', 'HEALTH_AND_FITNESS']
app_names = [f"App_{i}" for i in range(1, n_apps + 1)]

data = {
    'App': app_names,
    'Category': np.random.choice(categories, n_apps),
    'Rating': np.random.uniform(1.0, 5.0, n_apps).round(1),
    'Reviews': np.random.randint(10, 50000, n_apps),
    'Size': np.random.choice(['15M', '23M', '45M', '10M', '5.2M', '88M', '34M'], n_apps),
    'Installs': np.random.choice(['10,000+', '50,000+', '100,000+', '1,000,000+', '5,000,000+'], n_apps),
    'Type': np.random.choice(['Free', 'Paid'], n_apps, p=[0.85, 0.15]),
    'Price': np.random.choice([0.0, 0.99, 1.99, 2.99, 4.99], n_apps, p=[0.85, 0.05, 0.04, 0.03, 0.03])
}

df = pd.DataFrame(data)

print("--- Play Store Dataset Sample ---")
display(df.head())

# 2. Visualizations
plt.style.use('seaborn-v0_8-whitegrid')
fig = plt.figure(figsize=(15, 12))

# Graph 1: App Category Distribution (Bar Chart)
plt.subplot(2, 2, 1)
df['Category'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Number of Apps per Category', fontsize=12, fontweight='bold')
plt.xlabel('Category', fontsize=10)
plt.ylabel('Count', fontsize=10)
plt.xticks(rotation=45, ha='right')

# Graph 2: Distribution of App Ratings (Histogram)
plt.subplot(2, 2, 2)
sns.histplot(df['Rating'], bins=10, kde=True, color='green')
plt.title('Distribution of App Ratings', fontsize=12, fontweight='bold')
plt.xlabel('Rating', fontsize=10)
plt.ylabel('Frequency', fontsize=10)

# Graph 3: Free vs Paid Apps (Pie Chart)
plt.subplot(2, 2, 3)
type_counts = df['Type'].value_counts()
plt.pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', colors=['#ff9999','#66b3ff'], startangle=140)
plt.title('Free vs Paid Apps Proportion', fontsize=12, fontweight='bold')

# Graph 4: Rating vs Reviews (Scatter Plot)
plt.subplot(2, 2, 4)
sns.scatterplot(x='Rating', y='Reviews', hue='Type', data=df, palette='Set1', s=70)
plt.title('App Ratings vs Number of Reviews', fontsize=12, fontweight='bold')
plt.xlabel('Rating', fontsize=10)
plt.ylabel('Reviews', fontsize=10)

plt.tight_layout()
plt.show()
```
